Uma empresa contratou uma equipe (que são vocês) para analisar uma planilha de cancelamento dos planos dessa empresa. O objetivo da equipe é analisar a planilha para verificar o porque do cancelamento e o que pode ser feito a respeito para diminuir o número de cancelamentos.


A primeira coisa a ser feira e ler a planilha usando o pandas.

```python
import pandas as pd 

tabela = pd.read_csv("datasets/cancelamentos.csv")
display(tabela)
```

Vamos retirar a clouna **CustomerID** da consulta , pois ela noa e interessate para a analise.

Vamos analisar se essa tabela possui alguma valor nulo, e para isso vamos verificar o numero de linha da tabela.

```python
display(tabela.info())
```

Repare que o numero total de entradas nao bate com numero de valores nao nulas de cada uma das coluna, e nem o numero de linhas com valores nao-nulas e o o mesmo para cada uma das colunas.

Precisaremos eliminar os valores nulos da tabela.

```python
# eliminando os valores nulos
tabela = tabela.dropna()

display(tabela.info())
```

## Analise de dados
----


Agora que fizemos o tratamento de dados, a analise dde dados proprieamente dita vai cimerçat. Primeiramente, vamos verificar a qualidade de clientes que cancelaram o contrato versis a quantidade de clientes que ainda  estao com o contrato virgente.

```python
# retorna o total de cancelamnetos e o toltal de contratos ainda vigentes
display(tabela['cancelou'].value_counts())
```

A linha com o valor 1.0 representa os cancelamentos, equanto a linha com o valor 0.0 representa os contratos ainda virgentes.

Podemos pereceber que o numero de cancelamentos e maior do que o numero de contratos ainda viregentes,  oque significa que a empresa perdeu a maioria de seus clientes .A pergunta ahora e: por quê ?

Antes de respodemso a essa pergunta, vamos veridficar a porcetagem de cada um dos valores.

```python
display(tabela['cancelou'].value_counts(normalize=True).map("{:.2%}".format))
```

## Motando o perfil de cancelamento da empresa 
----


Vamos tentar descobrir qaul o perfil de cancelameto da empresa, e para isso camos analisar o s tipos de contratos existentes  na empresas na empresa com base no periodo de cada contrato.

```python
display(tabela['duracao_contrato'].value_counts())
display(tabela['duracao_contrato'].value_counts(normalize=True).map("{:.2%}".format))
```

Vamos agrupar as informações da coluna **dração_contract** e fazer  a media das informações .

```python
display(tabela.groupby('duracao_contrato').mean(numeric_only=True))
```

Repare que **EXATOS 100%**  dos cleintes que optaram pelo contrato mensal cancelaram o contrato logo, sabemos agora que esse ripo de contrato e problematico e portanto  precisamos  elima-lo
e exibir um relaroiro sem o contrasos messais.

```python
tabela = tabela[tabela['duracao_contrato'] != 'monthly']

display(tabela)
display(tabela['cancelou'].value_counts())
display(tabela['cancelou'].value_counts(normalize=True).map("{:.2%}".format))
```

A proporção de cancelamentos agora caiu para 46,05%. porem esse numero ainda e muito alto. Então precisamos ir mais a fundo  e ver onde epodemos eleiminar mais para dimininuir mais inda essa taxa.

## Graficos
----

Uma das melhoresformas de se analisar dados e atraves de graficos, pois fica muito mais facil visuzaluzar os dados o obter informações.

Para conseguirmos fazer isso, precisamos instalar um bliblioteca grafica para analaisar cada uma das colunas. Vamos a blibloteca platly , intalando-a atraves do comando pip install plotly no terminal.


```python
import plotly.express as px

for coluna in tabela.columns:
    grafico = px.histogram(tabela, x=coluna, color='cancelou', width=600)
    grafico.show()

   
```

Podemos verificar quea as taxas de cancelammentos sao de 100% para os clientes que:


1. Precisamosliigar para o call centert por mais de 5 vezes
2. Atrsaram o pagamento da assinatura por mais  de 20 dias.

Logo, vamos eleiminar da colsta todos os clientes que ligaram mais de 5 vezes para o call center, e tambem o pagamento por 10 dias ou mais .

```python
tabela = tabela[tabela["ligacoes_callcenter"] < 5]
tabela = tabela[tabela["dias_atraso"] <= 20]


display(tabela)
display(tabela['cancelou'].value_counts())
display(tabela['cancelou'].value_counts(normalize=True).map("{:.2%}".format))
```
